```python
import numpy as np

# Create a 2x3 starting array
# [[1, 2, 3],
#]
arr = np.array([[1, 2, 3], [4, 5, 6]])

print("Original Array:")
print(arr)

# 1. 'C' Order (Row-Major)
print("\n'C' Order (Row by Row):")
print(arr.ravel(order='C'))

# 2. 'F' Order (Column-Major)
print("\n'F' Order (Column by Column):")
print(arr.ravel(order='F'))

# 3. 'K' Order (Memory Layout Order)
print("\n'K' Order (True Memory Layout):")
print(arr.ravel(order='K'))
```

```python
import numpy as np

original = np.array([[1, 2], [3, 4]])

# --- Case 1: Using flatten() ---
flat_arr = original.flatten()
print('flatten Array\n')
flat_arr[0] = 99
print(flat_arr,'\n\n')
print('original\n\n',original,'\n\n')
# Output: 1 (The original is safe and unchanged!)

print('ravel Array\n')
# --- Case 2: Using ravel() ---
ravel_arr = original.ravel()
print(ravel_arr,'\n\n')
ravel_arr[0] = 99
print('original\n\n',original)
# Output: 99 (The original array was changed!)
```
